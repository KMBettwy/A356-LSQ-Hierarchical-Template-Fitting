```python
import matplotlib.pyplot as plt
import pandas as pd

rrab = pd.read_csv("bailey_data_ab.csv")
rrc  = pd.read_csv("bailey_data_c.csv")

plt.figure(figsize=(8,6))
plt.scatter(rrab['period'], rrab['amplitude'], color='steelblue', label='RRab', s=20)
plt.scatter(rrc['period'],  rrc['amplitude'],  color='tomato',   label='RRc',  s=20)

plt.title('Bailey Diagram for RR Lyrae Stars')
plt.xlabel('Period (days)')
plt.ylabel('Amplitude (mag)')
plt.legend()
plt.grid(True)
plt.show()
```
